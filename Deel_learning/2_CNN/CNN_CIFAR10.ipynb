```python
import tensorflow as tf                      # the deep learning library
from tensorflow.keras import layers, models  # layers = building blocks, models = to stack them
import matplotlib.pyplot as plt              # to show images
```

```python
# downloads the dataset: 60,000 training images + 10,000 test images
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# the labels are numbers 0-9, these are their names
class_names = ["T-shirt","Trouser","Pullover","Dress","Coat",
               "Sandal","Shirt","Sneaker","Bag","Ankle boot"]

print(x_train.shape)   # (60000, 28, 28) = 60000 images, each 28x28 pixels
```

```python
plt.imshow(x_train[0], cmap="gray")    # show the first image
plt.title(class_names[y_train[0]])     # show its label
plt.show()
```

```python
# pixels are 0-255, we make them 0-1 so the model learns easier
x_train = x_train / 255.0
x_test = x_test / 255.0

# a CNN wants images as (height, width, channels). Our images are grayscale = 1 channel
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
```

```python
model = models.Sequential([                          # layers stacked one after another

    layers.Input(shape=(28, 28, 1)),                 # input: a 28x28 grayscale image

    layers.Conv2D(32, 3, activation="relu"),         # 32 filters (3x3) find edges and simple patterns
    layers.MaxPooling2D(),                           # shrinks the image, keeps the strongest signals

    layers.Conv2D(64, 3, activation="relu"),         # 64 filters find more complex shapes
    layers.MaxPooling2D(),                           # shrink again

    layers.Flatten(),                                # turn the 2D result into one long list of numbers
    layers.Dense(64, activation="relu"),             # normal neural network layer to combine features
    layers.Dense(10, activation="softmax")           # 10 outputs = probability of each clothing class
])

model.summary()                                      # prints the structure
```

```python
# compile = tell the model HOW to learn
model.compile(
    optimizer="adam",                        # the method that updates the weights
    loss="sparse_categorical_crossentropy",  # measures how wrong the model is
    metrics=["accuracy"]                     # what we want to watch
)

# fit = actually train
model.fit(x_train, y_train, epochs=5)        # 5 epochs = go through all images 5 times
```

```python
# test on images the model has never seen
loss, accuracy = model.evaluate(x_test, y_test)
print("Test accuracy:", accuracy)

# predict one test image
prediction = model.predict(x_test[0:1])             # gives 10 probabilities
predicted_class = prediction.argmax()               # pick the biggest one

plt.imshow(x_test[0].reshape(28, 28), cmap="gray")
plt.title("Predicted: " + class_names[predicted_class] +
          " | Real: " + class_names[y_test[0]])
plt.show()
```
